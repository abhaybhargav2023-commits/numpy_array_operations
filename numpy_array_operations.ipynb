# Day 12: NumPy Array Operations

## Objective
Create two NumPy arrays and perform addition, subtraction, multiplication, and division.

## Learning Goals
- Create NumPy arrays using `np.array()`.
- Check array shapes.
- Perform element-wise mathematical operations.
- Understand vectorization in NumPy.
- Visualize array values and operation results using Matplotlib.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python

print("NumPy version:", np.__version__)
print("Libraries imported successfully!")
```

```python
array1 = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

array2 = np.array([
    [2, 4, 5],
    [5, 10, 10],
    [10, 8, 15]
])

print("Array 1:")
print(array1)

print("\nArray 2:")
print(array2)
```

```python
print("Shape of Array 1:", array1.shape)
print("Shape of Array 2:", array2.shape)

print("Dimensions of Array 1:", array1.ndim)
print("Dimensions of Array 2:", array2.ndim)

if array1.shape == array2.shape:
    print("Both arrays have the same shape.")
else:
    print("Array shapes are different.")
```

```python
addition = array1 + array2

print("Addition Result:")
print(addition)
```

```python
subtraction = array1 - array2

print("Subtraction Result:")
print(subtraction)
```

```python
multiplication = array1 * array2

print("Multiplication Result:")
print(multiplication)
```

```python
division = array1 / array2

print("Division Result:")
print(np.round(division, 2))
```

```python
operations = {
    "Addition": addition,
    "Subtraction": subtraction,
    "Multiplication": multiplication,
    "Division": division
}

for operation_name, result in operations.items():
    print(f"\n{operation_name}:")
    print(np.round(result, 2))
```

```python
fig, axes = plt.subplots(2, 3, figsize=(15, 9))

plots = [
    (array1, "Array 1", "viridis"),
    (array2, "Array 2", "YlOrRd"),
    (addition, "Addition", "Blues"),
    (subtraction, "Subtraction", "Greens"),
    (multiplication, "Multiplication", "Oranges"),
    (division, "Division", "Purples")
]

fig.suptitle(
    "Day 12: NumPy Array Operations Visualization",
    fontsize=18,
    fontweight="bold"
)

for ax, (data, title, color_map) in zip(axes.flat, plots):
    image = ax.imshow(data, cmap=color_map)

    ax.set_title(title, fontsize=13, fontweight="bold")
    ax.set_xlabel("Column Index")
    ax.set_ylabel("Row Index")

    ax.set_xticks(range(data.shape[1]))
    ax.set_yticks(range(data.shape[0]))

    for row in range(data.shape[0]):
        for col in range(data.shape[1]):
            value = data[row, col]

            ax.text(
                col, row, f"{value:.2f}" if title == "Division"
                else f"{value:g}",
                ha="center",
                va="center",
                color="black",
                fontsize=10,
                fontweight="bold"
            )

    fig.colorbar(image, ax=ax, fraction=0.046, pad=0.04)

plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()
```

```python
import time

# NumPy vectorized addition
start_time = time.perf_counter()
numpy_result = array1 + array2
numpy_time = time.perf_counter() - start_time

# Equivalent operation using Python loops
start_time = time.perf_counter()

loop_result = [
    [
        array1[i, j] + array2[i, j]
        for j in range(array1.shape[1])
    ]
    for i in range(array1.shape[0])
]

loop_time = time.perf_counter() - start_time

print("NumPy result:")
print(numpy_result)

print("\nPython loop result:")
print(np.array(loop_result))

print("\nResults match:", np.array_equal(numpy_result, loop_result))
print(f"NumPy execution time: {numpy_time:.8f} seconds")
print(f"Loop execution time:  {loop_time:.8f} seconds")
```
